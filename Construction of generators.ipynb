```python
# spanning tree algorithms
from collections import deque

# BFS algorithm
def BFS(adjacency, roots = [], depth_limit = -1):
    V = list(range(len(adjacency)))
    tree = adjacency*0
    Q = deque()
    depth = np.zeros(len(adjacency))+len(adjacency)+1
    components = np.zeros(len(adjacency))
    components_count = 0
    while len(V)>0:
        if len(roots)<=components_count:
            v = V.pop(0)
        else:
            v = roots[components_count]
            if v==-1:
                break
            V.remove(v)
        depth[v] = 0
        components[v] = components_count
        Q.append(v)
        while len(Q)>0:
            v = Q.popleft()
            if depth[v]==depth_limit:
                Q = deque()
                break
            for i in range(len(V)-1,-1,-1):
                if adjacency[v][V[i]]+adjacency[V[i]][v]==1:
                    Q.append(V[i])
                    depth[V[i]] = depth[v] + 1
                    components[V[i]] = components_count
                    tree[min(v,V[i])][max(v,V[i])] = 1
                    V.pop(i)
        components_count += 1
    return tree, depth, components

# DFS algorithm
def DFS_recursion(adjacency, V, tree, depth, v):
    for v2 in range(len(adjacency)):
        if adjacency[v][v2]+adjacency[v2][v]==1 and v2 in V:
            V.remove(v2)
            depth[v2] = depth[v] + 1
            tree[min(v,v2)][max(v,v2)] = 1
            DFS_recursion(adjacency, V, tree, depth, v2)
def DFS(adjacency):
    V = list(range(len(adjacency)))
    tree = adjacency*0
    depth = np.zeros(len(adjacency))
    while len(V)>0:
        v = V.pop(0)
        depth[v] = 0
        DFS_recursion(adjacency, V, tree, depth, v)
    return tree, depth

# BFS algorithm from graph center
def graph_center(adjacency):
    _,depth,components = BFS(adjacency)
    farthest1 = []
    for component in range(int(np.max(components))+1):
        depth_current = depth*(components==component)
        farthest1.append(np.argmax(depth_current))
    tree,depth,_ = BFS(adjacency,farthest1)
    farthest2 = []
    for component in range(int(np.max(components))+1):
        depth_current = depth*(components==component)
        farthest2.append(np.argmax(depth_current))
    graphcenters = []
    for farthest in farthest2:
        x = farthest
        while(depth[x]>depth[farthest]/2):
            for y in range(len(tree)):
                if tree[x][y]+tree[y][x]==1:
                    if depth[y]<depth[x]:
                        x = y
                        break
        graphcenters.append(x)
    return graphcenters
def BFS_graphcenter(adjacency):
    return BFS(adjacency, graph_center(adjacency))

# Kruskal's algorithm
def kruskal(adjacency, weights):
    edges = []
    edge_weights = np.zeros(int(np.sum(adjacency)))
    e = 0
    for i in range(len(adjacency)):
        for j in range(i, len(adjacency)):
            if adjacency[i][j] == 1:
                edges.append(np.array([i,j]))
                edge_weights[e] = weights[i,j]
                e += 1
    order = np.argsort(edge_weights)
    components = np.array(range(len(adjacency)))+1
    tree = adjacency*0
    for e in order:
        if components[edges[e][0]] != components[edges[e][1]]:
            tree[edges[e][0]][edges[e][1]] = 1
            component = min(components[edges[e][0]],components[edges[e][1]])
            components += -((components==components[edges[e][0]])*components[edges[e][0]]) - (components==components[edges[e][1]])*components[edges[e][1]] 
            components += (components==0)*component
    return tree

# create spanning tree
def spanning_tree(adjacency, choice):
    if choice==0:
        tree,depth = DFS(adjacency)
    if choice==1:
        tree,depth,_ = BFS(adjacency)
    if choice==2:
        tree,depth,_ = BFS_graphcenter(adjacency)
    return tree,depth
```

```python
# construct adjacency matrix of the graph and the dual graph
A_dual = np.zeros((len(triangles), len(triangles)))
A_dual_edges = A_dual + 0
A = np.zeros((len(vertices), len(vertices)))
A_edges = A + 0
edge_counter = 0
vertices_triangles = [[] for _ in range(len(vertices))]
for i in range(len(triangles)):
    for v in triangles[i]:
        vertices_triangles[v].append(i)
    for j in range(i,len(triangles)):
        edge = np.intersect1d(triangles[i],triangles[j])
        if len(edge)==2:
            edge_counter += 1
            A_dual[i][j] = 1
            A_dual_edges[i][j] = edge_counter
            A[edge[0]][edge[1]] = 1
            A_edges[edge[0]][edge[1]] = edge_counter
```

```python
# orient the mesh consistently
oriented = np.zeros(len(triangles))
def makeConsistent(k):
    for i in range(len(triangles)):
        if oriented[i]==0:
            if A_dual[i][k]+A_dual[k][i]==1:
                shared_edge, idx_i, idx_k = np.intersect1d(triangles[i],triangles[k], return_indices=True)
                if (idx_i[0]-idx_i[1])%3 == (idx_k[0]-idx_k[1])%3:
                   triangles[i][0], triangles[i][1] = triangles[i][1], triangles[i][0]
                oriented[i] = 1
                makeConsistent(i)
number_components = -1
connected_components = np.zeros(len(triangles))
for j in range(len(oriented)):
    if oriented[j]==0:
        oriented[j] = 1
        number_components += 1
        connected_components[j] = number_components
        makeConsistent(j)
# ensure normals point outside
triangle_norms = np.zeros(len(triangles))
for t in range(len(triangles)):
    triangle_norms[t] = np.linalg.norm((np.mean(vertices[triangles[t]], axis=0)))
for k in range(number_components+1):
    t_farthest = np.argmax(triangle_norms*(connected_components==k))
    vector1 = np.cross(vertices[triangles[t_farthest][1]]-vertices[triangles[t_farthest][0]],vertices[triangles[t_farthest][2]]-vertices[triangles[t_farthest][0]])
    vector2 = np.mean(vertices[triangles[t_farthest]], axis = 0)
    if np.dot(vector1, vector2) < 0:
        for i in range(len(triangles)):
            if connected_components[i] == k:
                triangles[i][0], triangles[i][1] = triangles[i][1], triangles[i][0]
```

```python
# construct spanning trees of the connected components of the dual graph
if choicealgo1 == 3:
    _,depth,_ = BFS_graphcenter(A)
    weights = A_dual*0
    for i in range(len(triangles)):
        for j in range(i,len(triangles)):
            if A_dual[i][j]==1:
                edge = np.intersect1d(triangles[i],triangles[j])
                weights[i][j] = -min(depth[edge[0]],depth[edge[1]])
    edges_D = kruskal(A_dual, weights)
else:
    edges_D,_ = spanning_tree(A_dual,choicealgo1)
```

```python
# construct the complement of the edge set of the dual graph
edges_b = A + 0
for i in range(len(edges_D)):
    for j in range(i,len(edges_D)):
        if edges_D[i][j]==1:
            e = np.intersect1d(triangles[i],triangles[j])
            edges_b[e[0]][e[1]] = 0

# construct set edges_star
Tree, depth = spanning_tree(edges_b,choicealgo2)
edges_star_matrix = edges_b - Tree
edges_star = []
for v in range(len(vertices)):
    for w in range(len(vertices)):
        if edges_star_matrix[v][w] == 1 :
            edges_star.append(np.sort([v,w]))
```

```python
# construction of circuits
s = []
for e in range(len(edges_star)) :
    s.append(deque())
    x,y = edges_star[e][0], edges_star[e][1]
    s[e].append(edges_star[e])
    while True:
        while depth[x]>depth[y]:
            for z in range(len(edges_b[x])): 
                if Tree[x][z]+Tree[z][x] == 1:
                    if depth[z]<depth[x]:
                        s[e].appendleft(np.array([z,x]))
                        x = z
                        break
        while depth[x]<=depth[y] and x != y:
            for z in range(len(edges_b[y])): 
                if Tree[y][z]+Tree[z][y] == 1:
                    if depth[z]<=depth[y]:
                        s[e].append(np.array([y,z]))
                        y = z
                        break
        if x == y:
            break
```

```python
# circuit optimization function
def circuit_optimizer(circuit,l):
    c = list(circuit)
    c_vertices = []
    for i in range(len(c)):
        c_vertices.append(c[i][0])
    e = 0
    while(e<len(c) and l<len(c)):
        x,y = c[e][0], c[(e+l-1)%len(c)][1]
        if e+l<len(c_vertices):
            path_current = c_vertices[e:e+l+1]
        else:
            path_current = c_vertices[e:]+c_vertices[:((e+l+1)%len(c_vertices))]
        vertices_avoid = [v for v in path_current if v not in [x,y]]
        A_path_search = A + 0
        A_path_search[vertices_avoid,:] = 0
        A_path_search[:,vertices_avoid] = 0
        path_tree, path_depth, _ = BFS(A_path_search, [x,-1], l-1)
        path_break = False
        for i in range(l-1):
            if c[(e+i)%len(c)][1] != c[(e+i+1)%len(c)][0]:
                path_break = True
        if path_break or (len(path_current) != len(set(path_current)) and len(path_current) != 3):
            e+=1
        elif (optimization_limit==-1 and path_depth[y] < l) or path_depth[y]==l-1 or (path_depth[y]==0 and l==2):
            path = deque()
            A_dual_test = A_dual + 0
            for k in range(l-1,-1,-1):
                remove = c[(e+k)%len(c)]
                A_dual_test -= (A_dual_edges==A_edges[np.min(remove)][np.max(remove)])
            while y!=x:
                for z in range(len(path_tree)):
                    if path_tree[y][z]+path_tree[z][y]==1 and path_depth[z]<path_depth[y]:
                        path.appendleft(np.array([z,y]))
                        A_dual_test -= (A_dual_edges==A_edges[min(y,z)][max(y,z)])
                        y = z
            _,_,test_components = BFS(A_dual_test)
            if np.max(test_components)-number_components == 1 or (l==2 and len(path)==0):
                for k in range(l-1,-1,-1):
                    if k < len(path):
                        c[(e+k)%len(c)] = path[k]
                    else:
                        c.pop((e+k)%len(c))
                        if e+k>len(c):
                            e = max(0,e-1)
                print(l, "replaced by", len(path))
                return c
            else:
                print("Failed attempt due to not homologous (tried to replace", l, "by", len(path),")")
                e+=1
        else:
            e += 1
    return c

def optimization_loop(circuit):
    print("Optimize circuit of length", len(circuit))
    l = 2
    while l<min(len(circuit), optimization_limit+1):
        c = circuit_optimizer(circuit,l)
        if len(c)<len(circuit):
            l = 1
            #visualization([circuit,c],0)
            circuit = c
        l += 1
    print("Length of new circuit:", len(circuit))
    return circuit
```

```python
# sumberging of cycles
# get normal vectors of all triangles and vertices
n_vec_triangles = np.zeros((len(triangles),3))
for i in range(len(triangles)):
    t = triangles[i]
    n_vec_triangles[i] = -np.cross(vertices[t[1]]-vertices[t[0]], vertices[t[2]]-vertices[t[0]])
n_vec_vertices = np.zeros((len(vertices),3))
for i in range(len(vertices)):
    for t in vertices_triangles[i]:
        n_vec_vertices[i] += n_vec_triangles[t]/len(vertices_triangles[i])
s_submerged = []
for cycle in s_new:
    submerge = []
    for e in cycle:
        submerge.append(vertices[e[0]]+n_vec_vertices[e[0]])
    s_submerged.append(submerge)
```

```python
# shifting of cycles
s_shifted = []
if not shifting:
    for cycle in s_new:
        shift = []
        for edge in cycle:
            shift.append(vertices[edge[0]])
        s_shifted.append(shift)
else:
    for c in range(len(s_new)):
        cycle, sequence = s_new[c], s_triangles[c]
        shift = []
        for i in range(len(cycle)):
            index = np.where(triangles[sequence[i]]==(cycle[i][0]))[0][0]
            shift.append((vertices[triangles[sequence[i]][(index-1)%3]] + vertices[triangles[sequence[i]][index]])/2)
            edge = np.delete(triangles[sequence][i], index)
            t = sequence[i]
            while len(np.intersect1d(edge, triangles[sequence[(i+1)%len(sequence)]]))!=2 :
                shift.append((vertices[edge[0]]+vertices[edge[1]])/2)
                t1,t2 = np.where(A_dual_edges==A_edges[np.min(edge), np.max(edge)])
                t = t1[0]+t2[0]-t
                edge = np.array([int(cycle[i][1]), int(np.setdiff1d(triangles[t],edge)[0])])
        s_shifted.append(shift)
```

```python
# visualize submerged and shifted cycles
for i in range(len(s)) :
    visualization([s_new[i]],0, "compare original cycle (purple) to submerged cycle (blue) and shifted cycle (yellow)", [s_submerged[i], s_shifted[i]])
```

```python
# calculate linking numbers
G = np.zeros((len(s),len(s)))
G_raw = G + 0
steps = 10
for a in range(len(G)):
    for b in range(len(G)):
        U,P = s_submerged[a], s_shifted[b]
        L = 0
        for i in range(len(U)):
            for j in range(len(P)):
                gradient = np.cross(U[(i+1)%len(U)]-U[i], P[(j+1)%len(P)]-P[j])
                x = np.linspace(U[i],U[(i+1)%len(U)],steps)
                y = np.linspace(P[j],P[(j+1)%len(P)],steps)
                Integral = 0
                for x_step in range(steps):
                    for y_step in range(steps):
                        Integral += np.dot(gradient, x[x_step]-y[y_step])/(4*np.pi*(np.linalg.norm(x[x_step]-y[y_step])**3))/(steps**2)
                L += Integral
        G_raw[a][b] = L
        G[a][b] = round(L)
print("linking numbers:")
print(G)
```

```python
# get basis of ker(G)
import sympy as sp
G_sp = sp.Matrix(G.astype(int))
basis_sp = G_sp.nullspace()
for i in range(len(basis_sp)):
    denominators = [x.q for x in basis_sp[i]]
    lcm = sp.ilcm(*denominators)
    basis_sp[i] = basis_sp[i]*lcm
basis = [np.array(v).astype(int).flatten() for v in basis_sp]
print("basis:", basis)
```

```python
# get bounding generators
s_bound_matrices = []
for basis_vector in basis:
    matrix = np.zeros((len(vertices), len(vertices)))
    for i in range(len(basis_vector)):
        if basis_vector[i] != 0:
            for edge in s_new[i]:
                matrix[np.min(edge)][np.max(edge)] += basis_vector[i]*((-1)**(edge[1]<edge[0]))
    s_bound_matrices.append(matrix)
s_bounding = []
def restore_cycle(generator, position, matrix, source):
    matrix_old = matrix+0
    for z in range(len(vertices)):
        y = generator[position][1]
        if (-1)**(z<y)*np.sign(matrix[min(y,z)][max(y,z)])==1:
            matrix[min(z,y),max(z,y)]-=np.sign(matrix[min(y,z)][max(y,z)])
            generator.insert(position+1,np.array([y,z]))
            if z == source:
                return True
            elif restore_cycle(generator, position+1, matrix, source):
                return True
            else:
                matrix = matrix_old+0
                generator.pop(position+1)
    return False
for matrix in s_bound_matrices:
    generator = []
    while(len(np.argwhere(matrix!=0))!=0):
        edge = np.argwhere(matrix!=0)[0]
        if np.sign(matrix[edge[0]][edge[1]]) == 1:
            x,y = edge[0],edge[1]
        else:
            x,y = edge[1],edge[0]
        matrix[edge[0],edge[1]]-=np.sign(matrix[edge[0],edge[1]])
        generator.append(np.array([x,y]))
        position = len(generator)-1
        restore_cycle(generator, position, matrix, generator[position][0])
        position = 0
        while(position<len(generator)):
            old_length = len(generator)
            restore_cycle(generator, position, matrix, generator[position][1])
            if old_length == len(generator):
                position+=1
    s_bounding.append(generator)
```
